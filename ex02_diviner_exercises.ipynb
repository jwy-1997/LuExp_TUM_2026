## Download data

```python
base_url = "https://luna1.diviner.ucla.edu/~jpierre/diviner/level4_raster_data"
longitude = 0

filename = f'diviner_tbol_snapshot_{str(longitude).zfill(3)}E.xyz'
retrieve_url = f'{base_url}/{filename}'

print("Downloading file from:", retrieve_url)
```

## Example code to plot a snapshot

```python
# Load data from txt file using numpy txt parser
data = np.loadtxt(filename)
data.shape
```

```python
data
```

```python
longitudes = np.arange(-179.75, 180, 0.5)
latitudes  = np.arange( -89.75,  90, 0.5)

raw_temperatures = data[:, 2]
```

```python
# Reshape temperatures so they are structured in a 2D (latitude, longitude) array
temperatures = raw_temperatures.reshape((len(longitudes), len(latitudes))).transpose()
temperatures = np.flip(temperatures, axis=0)
temperatures.shape
```

```python
plt.figure(figsize=(18, 6))
plt.imshow(temperatures, extent=[-180, 180, -90, 90])
plt.xlabel("Longitude (º)")
plt.ylabel("Latitude (º)")
plt.colorbar(label="Temperature (K)")
plt.show()
```
